# Color models

```python
import matplotlib.pyplot as plt
import numpy as np
import skimage.color as clr
import matplotlib.patches as patches
```

## RGB

```python
fig = plt.figure(figsize=(8,8))

# Axes for arrows
ax2 = fig.add_subplot()
ax2.set_axis_off()
ax2.set_xlim([-1,1])
ax2.set_ylim([-1,1])

# Arrow
arrow_kw = dict(arrowstyle="Simple, tail_width=0.5, head_width=4, head_length=8",
                color="k")
text_kw = dict(size=20)

# Arrow for red
ax2.text(-1, -.37, "Red", rotation=23, **text_kw)
a = patches.FancyArrowPatch((.03,0), (-1, -.44), **arrow_kw)
ax2.add_patch(a)

# Arrow for green
ax2.text(.78, -.38, "Green", rotation=-23, **text_kw)
a = patches.FancyArrowPatch((.03,0), (1, -.43), **arrow_kw)
ax2.add_patch(a)

# Arrow for blue
ax2.text(-.08, .85, "Blue", rotation=90, **text_kw)
a = patches.FancyArrowPatch((.03,0), (.03, 1), **arrow_kw)
ax2.add_patch(a)

# Axe for the cube
ax1 = fig.add_subplot(projection='3d', facecolor=(0,0,0,0), proj_type='ortho')
ax1.view_init(elev=30, azim=45)
ax1.set_axis_off()

# Number of facet
N = 50

for side in range(3):

    # The two axis of the cube side
    X1 = np.linspace(0, 1, N+1)
    X2 = np.linspace(0, 1, N+1)
    X1, X2 = np.meshgrid(X1, X2)

    # The constant axis
    X0 = np.ones((N+1,N+1)) * 1

    # The colors of the cube side
    C1 = np.linspace(0, 1, N)
    C2 = np.linspace(0, 1, N)
    C1, C2 = np.meshgrid(C1, C2)

    # The constant color
    C0 = np.ones((N,N)) * 1

    # Order the axis and colors depending on the side to be displayed
    if side == 0:
        X, Y, Z = X0, X1, X2
        colors = np.array([C0, C1, C2])
    if side == 1:
        X, Y, Z = X1, X0, X2
        colors = np.array([C1, C0, C2])
    if side == 2:
        X, Y, Z = X1, X2, X0
        colors = np.array([C1, C2, C0])

    # Order the array colors to fit with X and Y
    colors = np.moveaxis(colors, 0, -1)

    # Plot
    surf = ax1.plot_surface(X, Y, Z, facecolors=colors,
                           linewidth=2, antialiased=True, shade=False, alpha=1)
    
# Save and show
plt.savefig("model-rgb.svg", transparent=True, bbox_inches="tight")
plt.show()
```

## HSV

### Functions

```python
def hsv_cylinder(Hlen, Vlen):
    
    # Cylinder of HSV color space

    # H, S, V grid
    H = np.linspace(0, np.pi*2, Hlen)
    S = np.ones((Vlen, Hlen)) * 1
    V = np.linspace(0, 1, Vlen)
    H, V = np.meshgrid(H, V)
    
    # X, Y, Z grid
    X = np.cos(H)
    Y = np.sin(H)
    Z = V

    # Color grid
    CH = np.linspace(0, 360, Hlen)
    CS = np.ones((Vlen,Hlen)) * 1
    CV = np.linspace(0, 1, Vlen)
    CH, CS = np.meshgrid(CH, CS)

    # Conversion HSV to RGB
    HSV = np.array((H.ravel()/(2*np.pi), S.ravel(), V.ravel()))
    HSV = np.moveaxis(HSV, 0, -1)
    RGB = clr.hsv2rgb(HSV)
    R = np.reshape(RGB[:,0], (Vlen, Hlen))
    G = np.reshape(RGB[:,1], (Vlen, Hlen))
    B = np.reshape(RGB[:,2], (Vlen, Hlen))

    # Color output
    colors = np.array([R, G, B])
    colors = np.moveaxis(colors, 0, -1)

    return X, Y, Z, colors
```

```python
def hsv_disk(Hlen, Slen):
        
    # Top disk of HSV color space

    # H, S, V grid
    H = np.linspace(0, 2*np.pi, Hlen)
    S = np.linspace(0, 1, Slen)
    V = np.ones((Slen, Hlen)) * 1
    H, S = np.meshgrid(H, S)
    
    # X, Y, Z grid
    X = S * np.cos(H)
    Y = S * np.sin(H)
    Z = np.ones((Slen, Hlen)) * 1
    
    # Color grid
    CH = np.linspace(0, 360, Hlen)
    CS = np.linspace(0, 1, Slen)
    CH, CS = np.meshgrid(CH, CS)
    CV = np.ones((Slen,Hlen)) * 1
    
    # Conversion HSV to RGB
    HSV = np.array((H.ravel()/(2*np.pi), S.ravel(), V.ravel()))
    HSV = np.moveaxis(HSV, 0, -1)
    RGB = clr.hsv2rgb(HSV)
    R = np.reshape(RGB[:,0], (Slen, Hlen))
    G = np.reshape(RGB[:,1], (Slen, Hlen))
    B = np.reshape(RGB[:,2], (Slen, Hlen))
    
    # Color output
    colors = np.array([R, G, B])
    colors = np.moveaxis(colors, 0, -1)

    return X, Y, Z, colors
```

### Main

```python
fig = plt.figure(figsize=(8,8))

# 3D Axes
ax1 = fig.add_subplot(projection='3d', proj_type='ortho')
ax1.view_init(elev=30, azim=45)
ax1.set_axis_off()

# HSV color space
Hlen, Slen, Vlen = 100, 20, 50
kw = dict(linewidth=10, antialiased=True, shade=False, alpha=1)
X, Y, Z, colors = hsv_cylinder(Hlen, Vlen)
ax1.plot_surface(X, Y, Z, facecolors=colors, rcount=Vlen, ccount=Hlen, **kw)
X, Y, Z, colors = hsv_disk(Hlen, Slen)
ax1.plot_surface(X, Y, Z, facecolors=colors, rcount=Slen, ccount=Hlen, **kw)

# 2D axes (for arrows)
ax2 = fig.add_subplot()
ax2.set_axis_off()
ax2.set_xlim([-1,1])
ax2.set_ylim([-1,1])

# Styles
arrow_kw = dict(arrowstyle="Simple, tail_width=0.5, head_width=4, head_length=8",
                color="k")
text_kw = dict(size=20)

# Arrow for hue
ax2.text(-.73, .05, "Hue", rotation=-50, **text_kw)
a = patches.FancyArrowPatch((-.6, .37), (-.16, .0), **arrow_kw, connectionstyle="Angle3")
ax2.add_patch(a)
    
# Arrow for saturation
ax2.text(.07, .32, "Saturation", rotation=-20, **text_kw)
a = patches.FancyArrowPatch((.7, -0.5), (.7, .5), **arrow_kw)
ax2.add_patch(a)

# Arrow for value
ax2.text(.72, -.1, "Value", rotation=-90, **text_kw)
a = patches.FancyArrowPatch((.03, .44), (.48, .27), **arrow_kw)
ax2.add_patch(a)

# Save and show
plt.savefig("model-hsv.svg", transparent=True, bbox_inches="tight")
plt.show()
```
