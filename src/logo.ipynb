# Logo

```python
import numpy as np
import matplotlib.pyplot as plt
import skimage.transform as tf
```

```python
def logo(W, H, back="tranparent", scale=4):
    """Logo de la synthèse additive.
    W, H  : dimensions de l'image en pixel.
    back  : type de fond (blanc, noir, gradient...). Transparent par défaut.
    scale : échelle de redimensionnement (pour éviter le crénelage).
    """
    
    # Dimension de l'image
    W, H = W*scale, H*scale

    # Rayon des cercles
    r = np.min((W, H))/3

    # Distance des cercles au centre
    d = r*.5

    # Centres des cercles
    thetas = 2 * np.pi / 3 * np.arange(3)
    cx = W/2 + d * np.sin(thetas)
    cy = H/2 - d * np.cos(thetas) + H*.03

    # Construction de l'image
    img = np.zeros((H,W,4))
    w, h = np.meshgrid(np.arange(W), np.arange(H))
    for b in range(3):
        img[:,:,b] = (w-cx[b])**2 + (h-cy[b])**2 < r**2

    # Fond de l'image
    s = np.sum(img, axis=2)
    x, y = np.where(s==0)
    if back == "white":
        img[x,y,:] = 1
        img[:,:,3] = 1
    elif back == "color-gradient":
        grad = np.ones((H,W,3))
        grad[:,:,0] = h/H
        grad[:,:,1] = (W-w)/W
        grad[:,:,2] = w/W
        img[x,y,:] = grad[x,y,:]
        img[:,:,3] = 1
    elif back == "gradient":
        g = np.linspace(.4, .2, H)
        g = np.tile(g,(W,1)).T
        grad = np.ones((H,W,3))
        grad[:,:,0] = g
        grad[:,:,1] = grad[:,:,0]
        grad[:,:,2] = grad[:,:,0]
        img[x,y,:3] = grad[x,y,:]
        img[:,:,3] = 1
    elif back == "black":
        img[x,y,:] = 0
        img[:,:,3] = 1
    else:
        x, y = np.where(s>0)
        img[x, y, 3] = 1

    # Redimensionnement
    img = tf.rescale(img, (1/scale, 1/scale, 1), order=2, anti_aliasing=True)

    return img
```

```python
# Logo Moodle
f = logo(W=400, H=190, back="gradient", scale=4)
plt.imsave("logo-bip-moodle.png", f)
```

```python
# Logo sur la page du book
f = logo(W=50, H=50, back="transparent", scale=4)
plt.imsave("logo-bip-site.png", f)
```
