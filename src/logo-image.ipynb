```python
def logo(dwg, w, h, s):
    """Logo des trois pixels."""

    # Rouge : 0, s/2
    # Vert  : s/2, 0
    # Bleu  : s*2/3, s*2/3
    
    x, y = w/2-20, h/2-20
    alpha = .5

    dwg.add( dwg.rect(insert=(0,0), size=(w,h), fill="white") )

    dwg.add( dwg.rect(insert=( x+0,  y+s/2 ), size=(s,s), fill="red"), blend='screen')
    dwg.add( dwg.rect(insert=( x+s/2, y+0  ), size=(s,s), fill="green") )

    return dwg

# Dimensions de l'image
w, h = 500, 300

# Dimension des pixels/carrés
s = 100

dwg = svgwrite.Drawing('logo-bip-site.svg', size=(f'{w}px', f'{h}px'))
dwg = logo(dwg, w, h, s)
dwg.save()

```

```python
import svgwrite

# 1. Création du dessin SVG (400x400 pixels)
dwg = svgwrite.Drawing('logo-bip-site.svg', size=('400px', '400px'))

# 2. Arrière-plan noir
dwg.add(dwg.rect(insert=(0, 0), size=('100%', '100%'), fill='black'))

# 3. Définition des filtres d'addition pour chaque couche
# Filtre 1 : Fusionne le Vert avec le Rouge sous-jacent
filter_green = dwg.filter(id='blend_green', x='0%', y='0%', width='100%', height='100%')
filter_green.add(dwg.element('feBlend', in_='SourceGraphic', in2='BackgroundImage', mode='screen'))
dwg.defs.add(filter_green)

# Filtre 2 : Fusionne le Bleu avec les couches précédentes
filter_blue = dwg.filter(id='blend_blue', x='0%', y='0%', width='100%', height='100%')
filter_blue.add(dwg.element('feBlend', in_='SourceGraphic', in2='BackgroundImage', mode='screen'))
dwg.defs.add(filter_blue)

# 4. Groupe activant la capture d'arrière-plan (enable-background)
groupe = dwg.g(style="enable-background: new;")

# Rayon des disques
r = 80

# Disque 1 : Rouge (servira de base)
groupe.add(dwg.circle(center=(200, 155), r=r, fill='rgb(255, 0, 0)'))

# Disque 2 : Vert (fusionné via feBlend)
groupe.add(dwg.circle(center=(161, 222), r=r, fill='rgb(0, 255, 0)', filter='url(#blend_green)'))

# Disque 3 : Bleu (fusionné via feBlend)
groupe.add(dwg.circle(center=(239, 222), r=r, fill='rgb(0, 0, 255)', filter='url(#blend_blue)'))

# 5. Sauvegarde
dwg.add(groupe)
dwg.save()
```

```python
# Taille de l'image
W, H = 50, 50

# Nombre de points du sinus cardinal
N = 70

# Sinus cardinal
n = np.linspace(-1, 1, N)
x = np.linspace(0, W, N)
y = np.sinc(5*n)
y = 30 - y*15

# Liste des coordonnées du sinus cardinal
points = ""
for k in range(N):
    if k==0:
        points += "M "
    else:
        points += "L "
    points += f"{x[k]} {y[k]}"

# Initialisation du SVG
dwg = svgwrite.Drawing('logo-signal2-site.svg', size=(f'{W}px', f'{H}px'))

# # Rectangle blanc
# dwg.add( dwg.rect(insert=(0, 0), size=(f'{W}', f'{H}'), fill="red") )

# Masque
mask = dwg.mask(id="themask")
mask.add(
    dwg.circle(center=(W/2, H/2), r=.48*W, fill="white")
)
dwg.defs.add(mask)

# Cercle
dwg.add(
    dwg.circle(center=(W/2, H/2), r=.48*W, stroke="black", fill="white", stroke_width=1.5)
)

# Dessin du sinus cardinal
dwg.add(
    dwg.path(d=points, stroke="black", fill="none", stroke_width=2, mask="url(#themask)")
)

# Enregistrement
dwg.save()
```

```python
# Taille de l'image
W, H = 540, 190

# Nombre de points du sinus cardinal
N = 80

# Largeur du sinus cardinal
w = 200

# Sinus cardinal
n = np.linspace(-6, 6, N)
x = np.linspace(W/2-w/2, W/2+w/2, N)
y = 110 - 70 * np.sinc(n)

# Liste des coordonnées du sinus cardinal
points = ""
for k in range(N):
    p = "M" if k==0 else "L"
    points += f"{p} {x[k]} {y[k]}"

# Initialisation du SVG
dwg = svgwrite.Drawing('logo-signal2-moodle.svg', size=(f'{W}px', f'{H}px'))

# Gradient de couleur
gradient = dwg.linearGradient((0, 0), (1, 0), id="gradient")
gradient.add_stop_color(offset='0%',   color='rgb(0, 176, 50)', opacity=None)
gradient.add_stop_color(offset='100%', color='rgb(0, 219, 51)', opacity=None)
dwg.defs.add(gradient)
dwg.add(
    dwg.rect(insert=(0, 0), size=(f'{W}', f'{H}'), fill="url(#gradient)")
)

# # Cercle
# dwg.add(
#     dwg.circle(center=(W/2, H/2), r=.4*H, stroke="white", fill="none", stroke_width=5)
# )

# Masque
mask = dwg.mask(id="themask")
mask.add(
    dwg.circle(center=(W/2, H/2), r=.4*H, fill="white")
)
dwg.defs.add(mask)

# Dessin du sinus cardinal
dwg.add(
    dwg.path(d=points, stroke="white", fill="none", stroke_width=5, mask="url(#themask)")
)

# Enregistrement
dwg.save()
```
