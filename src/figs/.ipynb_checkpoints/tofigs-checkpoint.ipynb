# Format images to be used in the book

```python
import numpy as np
import skimage.io as io
from skimage.transform import resize
import matplotlib.pyplot as plt
```

```python
def format_image(filename, crop_origin, crop_size, size, newname):
    
    # Load image
    img = io.imread(filename)
    
    # Crop image
    r, c = crop_origin
    img = img[r:(r+crop_size),c:(c+crop_size)]
    
    #Resize image
    img = resize(img, size, anti_aliasing=True)
    
    # # Display image
    # plt.figure(figsize=(8,8))
    # plt.imshow(img, cmap='gray')
    # plt.show()
    
    # Save image
    io.imsave(newname, img)

    return img
```

```python
img = format_image('337294main_pg62_as11-40-5903_full.jpg', (100, 820), 2600, (512,512,3), 'aldrin.png')
```

```python
# https://unsplash.com/photos/0PUoXmsCSDQ/
    
img = io.imread('irene-kredenets-0PUoXmsCSDQ-unsplash.jpg')
r, c, N = 670, 400, 1100
img = img[r:(r+N),c:(c+N)]
img = resize(img, (512,512,3), anti_aliasing=True)

plt.figure(figsize=(8,8))
plt.imshow(img, cmap='gray')
plt.show()
```

```python
# https://nssdc.gsfc.nasa.gov/imgcat/html/object_page/lo1_m31.html

img = io.imread('lo1_m31.jpg')
r, c, N = 20, 20, 520
img = img[r:(r+N),c:(c+N)]
img = resize(img, (512,512,3), anti_aliasing=True)

plt.figure(figsize=(8,8))
plt.imshow(img, cmap='gray')
plt.show()
```

```python
img = np.zeros((128, 128))

m, n = np.meshgrid(range(128),range(128))

spots = [
    [ 40, 100, 3, 1.00],
    [ 83,  65, 3, 0.10],
    [ 17,  32, 3, 0.10],
    [ 34,  79, 10, 0.05],
    [ 68,  12, 2, 0.2],
]
for spot in spots:
    x, y, sig, a = spot
    img += a * np.exp(-(m-x)**2/2/sig**2-(n-y)**2/2/sig**2)

# Affichage
plt.figure(figsize=(8, 8))
plt.imshow(img, cmap="gray")
plt.show()
io.imsave("spots.png", img)
```
