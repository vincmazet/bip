```python
import numpy as np
from skimage import io
from skimage.color import rgb2gray
from skimage.feature import SIFT, match_descriptors, plot_matched_features
import matplotlib.pyplot as plt
```

```python
# Chargement des images
img1 = io.imread("../../src/figs/Strasbourg1950.jpg")
img2 = io.imread("../../src/figs/Strasbourg2024.jpg")

# Affichage
fig = plt.figure(figsize=(7,4))
ax = fig.add_subplot([.1, .1, .4, 1])
ax.imshow(img1)
ax.axis('off')
ax = fig.add_subplot([.6, .1, .4, 1])
ax.imshow(img2)
ax.axis('off')

# Annotations
markers = [
    [350, 425, -295, 380, 'C0'],
    [275, 280, -420, 270, 'C1'],
    [180, 190, -540, 220, 'C2'],
    [455, 265, -260, 190, 'C3'],
    [120, 450, -510, 480, 'C4'],
]
for x0, y0, x1, y1, clr in markers:
    ax.annotate("", (x0,y0), xytext=(x1,y1), arrowprops=dict( arrowstyle="-", lw=2, color=clr))

# Finalisation de l'affichage et enregistrement
plt.savefig("feature-matching.svg", bbox_inches="tight")
plt.show()
```
