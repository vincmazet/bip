# Corners

```python
import matplotlib.pyplot as plt
import skimage.io as io
from skimage.feature import corner_harris

plt.style.use('../_static/bip.mplstyle')
plt.rcParams['image.cmap'] = 'gray'
```

```python
# Image
img = io.imread('moulinsart.png')

# Harris detector
res = corner_harris(img, method='k', k=0.05)

# # Display
# fig, axs = plt.subplots(1, 4, figsize=(20,8))
# axs[0].imshow(img)
# axs[0].set_title('Original image')
# axs[1].imshow(res<-.1)
# axs[1].set_title('$R<-0.1$')
# axs[2].imshow((res>-.1)*(res<.5))
# axs[2].set_title('$R>-0.1$ and $R<0.5$')
# axs[3].imshow(res>.5)
# axs[3].set_title('$R>0.5$')
# for i in range(4):
#     axs[i].set_xticks([])
#     axs[i].set_yticks([])
# plt.savefig('harris-detector.svg', bbox_inches='tight')
# plt.show()

# Display
c = .43
fig = plt.figure(figsize=(15,10))

ax = fig.add_axes([.15, .5, c, c])
ax.imshow(img)
ax.set_title('Original image')

ax = fig.add_axes([0.5, .5, c, c])
im = ax.imshow(res, "hsv")
ax.set_title('$R$')

ax = fig.add_axes([0, 0, c, c])
ax.imshow(res<-.1)
ax.set_title('$R<-0.1$')

ax = fig.add_axes([.33, 0, c, c])
ax.imshow((res>-.1)*(res<.5))
ax.set_title('$R>-0.1$ and $R<0.5$')

ax = fig.add_axes([.66, 0, c, c])
ax.imshow(res>.5)
ax.set_title('$R>0.5$')

for ax in fig.axes:
    ax.set_xticks([])
    ax.set_yticks([])
    

ax = fig.add_axes([0.88, .5, .03, c])
fig.colorbar(im, cax=ax, orientation='vertical')

plt.savefig('harris-detector.svg', bbox_inches='tight')
plt.show()
```
