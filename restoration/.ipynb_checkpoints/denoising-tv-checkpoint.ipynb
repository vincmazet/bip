# Débruitage par régularisation TV

```python
from numpy import arange
from skimage.io import imread, imsave
from skimage.color import rgb2gray
from matplotlib.pyplot import imshow, show
from skimage.restoration import denoise_tv_chambolle
```

```python
# Zoom
z1, z2 = 122, 162
```

```python
x = imread("lena-psnr-20.png")
x = rgb2gray(x)
M, N = x.shape
imshow(x[z1:z2,z1:z2], cmap="gray")
show()
```

```python
for lbd in [10, 1000]:
    x_tv = denoise_tv_chambolle(x, weight=lbd/1000)
    x_tv = x_tv[z1:z2,z1:z2]
    imshow(x_tv, cmap="gray")
    show()
    imsave("lena-tv-" + str(lbd) + "-zoom.png", x_tv)
```
