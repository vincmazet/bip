# Convolution

```python
import numpy as np
import matplotlib.pyplot as plt
plt.style.use('../_static/bip.mplstyle')
```

```python
# Figure
w, h = 15.1, 7
debug = False
z = .7
fig, ax = plt.subplots(1, 1, figsize=(w*z+1,h*z+1))
ax.set_aspect('equal')
ax.axis('off')
ax.set_xlim(0, w)
ax.set_ylim(0, h)
ax.yaxis.set_inverted(True)
if debug:
    p = plt.Rectangle((0,0), w, h, fill=False, edgecolor="tab:cyan", linewidth=2)
    ax.add_patch(p)

def block(dx, dy, x, y, txt, color="gray", facecolor="white"):
    p = plt.Rectangle((dx+x, dy+y), 1, 1, facecolor=facecolor, edgecolor="black")
    ax.add_patch(p)
    plt.text(dx+x+.5, dy+y+.5, txt, ha="center", va="center", color=color)

# Couleurs
colors = [ "#03045e", "#023e8a", "#0077b6","#0096c7", "#00b4d8", "#48cae4", "#90e0ef", "#ade8f4", "#caf0f8" ]
colors = ["#2b0000","#4f0000","#740000","#980000","#b50000","#d30000","#eb1d1d","#f50f0f","#ff0000"]
colors = ["#370617","#6a040f","#9d0208","#d00000","#dc2f02","#e85d04","#f48c06","#faa307","#ffba08"]

# Image f
S = 5
y, x  = np.mgrid[0:S, 0:S]
x = x.ravel()
y = y.ravel()
i = 0
for k in range(S**2):
    if k == 6:
        color = colors[4]
    else:
        color = "gray"
    txt = f"$f_{{{y[k]},{x[k]}}}$"
    fc = "white"
    block(0, 0, x[k], y[k], txt, color=color, facecolor=fc)

# Image g
S = 5
y, x  = np.mgrid[0:S, 0:S]
x = x.ravel()
y = y.ravel()
i = 0
for k in range(S**2):
    txt = f"$g_{{{y[k]},{x[k]}}}$"
    if x[k]<3 and y[k]<3:
        color = colors[i]
        i += 1
    else:
        color = "gray"
    block(6, 0, x[k], y[k], txt, color=color)

# Image h
S = 3
y, x  = np.mgrid[0:S, 0:S]
x = x.ravel()
y = y.ravel()
for k in range(S**2):
    txt = f"$h_{{{y[k]-1},{x[k]-1}}}$"
    block(12, 1, x[k], y[k], txt, color=colors[-(k+1)])

# Symboles mathématiques
plt.text( 5+.5, 2+.5, "=", ha="center", va="center", color="k", fontsize=25)
plt.text(11+.5, 2+.5, "*", ha="center", va="center", color="k", fontsize=20)

# Équation
p = plt.Rectangle((0,5.5), 15, 2, facecolor="w")
ax.add_patch(p)
plt.text(1, 6, "$f_{1,1}$", color=colors[4])
plt.text(1.7, 6, "=")
# S = 3
# y, x  = np.mgrid[0:S, 0:S]
# x = x.ravel()
# y = y.ravel()
xo, yo = 0.15, 6
for k in range(S**2):
    xo += 2
    if xo > 14:
        xo = 2.15
        yo += .7
    # plt.text(1.45+k*2, 6, F"$g_{{{y[k]},{x[k]}}} \\!\\cdot\\! h_{{{S-y[k]-2},{S-x[k]-2}}}$", color=colors[k])
    plt.text(xo, yo, F"$g_{{{y[k]},{x[k]}}} \\!\\cdot\\! h_{{{S-y[k]-2},{S-x[k]-2}}}$", color=colors[k])
    if k < S**2-1:
        plt.text(xo+1.5, yo, "+")

# Enregistrement et affichage
plt.savefig("convolution.svg", dpi=200, facecolor='None', edgecolor='None', transparent=False, bbox_inches='tight')    
plt.show()
```
