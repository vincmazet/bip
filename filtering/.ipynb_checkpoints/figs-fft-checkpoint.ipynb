```python
import numpy as np
import skimage.io as io
import matplotlib.pyplot as plt
plt.style.use('book/_static/bip.mplstyle')

# Supprime les warnings indiquant les log(0)
import warnings
warnings.filterwarnings('ignore')
```

```python
# Taille de l'image
N = 16

# Pixels (dans le domaine spectral) à considérer
px = (
    (( 0, 0, 1), ),
    (( 0, 1, 1), ),
    (( 0, 1,-1), ),
    # (( 0, 2, 1), ),
    (( 1, 0, 1), ),
    (( 1, 1, 1), ),
    (( 2, 2, 1), ),
    (( 2, 0, 1), ( 0, 2, 1), ),
)
K = len(px)

# Affichage
for k in range(K):

    # Initialisation de la FFT et ajout des pixels
    X = np.zeros((N,N), dtype="complex")
    for i1, j1, s in px[k]:
        i2 = N - i1 if i1>0 else 0
        j2 = N - j1 if j1>0 else 0
            
        # FFT (non shiftée)
        X[i1,j1] += .5e3 if s==1 else  .5e3*1j
        X[i2,j2] += .5e3 if s==1 else -.5e3*1j
    
    # FFT inverse
    x = np.fft.ifft2(X)
    x = np.real(x)

    # FFT shiftée
    X = np.fft.fftshift(X)

    # Affichage
    fig, axs = plt.subplots(1, 3, figsize=(9,3))
    
    # Affichage de l'image de la base
    axs[0].imshow(x, cmap='gray', vmin=-4, vmax=4)
    axs[0].set_xticks([0, 15])
    axs[0].set_yticks([0, 15])
    
    # Affichage du module de la FFT
    axs[1].imshow(np.abs(X), cmap='gray', vmin=0, vmax=.5)
    axs[1].set_xticks([0, 8, 15])
    axs[1].set_xticklabels([-8, 0, 7])
    axs[1].set_yticks([0, 8, 15])
    axs[1].set_yticklabels([-8, 0, 7])
    
    # Affichage de la phase de la FFT
    axs[2].imshow(np.angle(X), cmap='gray', vmin=-np.pi/2, vmax=np.pi/2)
    axs[2].set_xticks([0, 8, 15])
    axs[2].set_xticklabels([-8, 0, 7])
    axs[2].set_yticks([0, 8, 15])
    axs[2].set_yticklabels([-8, 0, 7])

    # Enregistrement et affichage
    plt.savefig(f"fourier-{k}.png", dpi=200, facecolor='None', edgecolor='None', transparent=False, bbox_inches='tight')    
    plt.show()
```
