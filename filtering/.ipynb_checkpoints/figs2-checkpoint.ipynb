```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.fft import fft2, fftshift, ifft2, ifftshift

# import matplotlib as mpl
plt.style.use('../_static/bip.mplstyle')
```

```python
def display(f=None, F=None, filename=None):
    
    # Affiche une image f et sa DFT F (module et argument)
    
    if f is None:
        f = np.real( ifft2(ifftshift(F)) )
    if F is None:
        F = fftshift(fft2(f))

    (M, N) = f.shape
    
    fig, axs = plt.subplots(1, 3, figsize=(15,5))
    axs[0].imshow(f, cmap="gray")
    axs[0].set_xlabel("Image")
    axs[1].imshow(np.log(np.absolute(F)+1e-12), cmap="gray", extent = [-M/2 , M/2, -N/2, N/2])
    axs[1].set_xlabel("Amplitude of the DFT")
    axs[2].imshow(np.angle(F), cmap="gray", extent = [-M/2 , M/2, -N/2, N/2])
    axs[2].set_xlabel("Phase of the DFT")

    if filename:
        plt.savefig(filename)

# # Test
# import skimage.io as io
# f = io.imread("../_static/figs/squirrel.png")
# display(f=f)
# display(F=F)
```

```python
# Construction d'une TFD dans le domaine de Fourier
def wave(N, rho, theta, c, ampl):
    
    F = np.zeros((N,N), dtype="complex")
    
    e = rho * np.exp(1j*theta*np.pi)
    a = int(np.round(np.real(e)))
    b = int(np.round(np.imag(e)))
    
    F[N//2-a,N//2+b] = ampl * 1/2 * (1 if c==0 else +1j)
    F[N//2+a,N//2-b] = ampl * 1/2 * (1 if c==0 else -1j)

    return F

# Construction des trois TFD (simples oscillations)
N = 64
idx = [
    [9,   0, 0, 1],
    [9, 1/5, 1, 1],
    [2, 1/4, 0, 3],
]
I = len(idx)

G = np.zeros((N,N), dtype="complex")
for i in range(I):

    rho, theta, c, ampl = idx[i]
    F = wave(N, rho, theta, c, ampl)
    display(F=F)
    G += F
    
# Affichage des TFD et de leur somme
display(F=G)
```
