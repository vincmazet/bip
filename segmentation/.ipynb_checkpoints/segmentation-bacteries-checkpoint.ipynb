# Segmentation de bactéries

* Image : PIA17369_hires.jpg (NASA/JPL-Caltech) [https://www.jpl.nasa.gov/spaceimages/details.php?id=PIA17369]

Bacterial cells of the recently discovered species Tersicoccus phoenicis, found in only two places:
clean rooms in Florida and South America where spacecraft are assembled for launch.
The monitoring provides an indication of what species might get into space aboard a spacecraft.

```python
from skimage.io import imread, imsave
from skimage.color import rgb2gray
from skimage import img_as_ubyte
from matplotlib.pyplot import imshow, hist, show
```

```python
# Chargement
f = imread('PIA17369_hires.jpg')

# Mise en forme
f = rgb2gray(f)
f = f[0:500, 0:500]
f = img_as_ubyte(f)
print("%s, [%d, %d]" % (f.dtype, f.min(), f.max()))
```

```python
# Enregistrement
imshow(f, cmap="gray")
imsave("bacteries.png", f)
```

```python
# Histogramme
hist(f.ravel(), bins=256);
```

```python
# Seuillage
g = (f<50) * 255
imshow(g, cmap="gray")
imsave("bacteries-seg.png", g)
```
